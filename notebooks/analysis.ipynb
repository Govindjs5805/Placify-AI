```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set visualization style
sns.set_theme(style="whitegrid")

# 1. Load the dataset
df = pd.read_csv('../dataset/students.csv')  # Adjust path if needed

# 2. Basic Inspection
print("--- First 5 Rows ---")
print(df.head())

print("\n--- Data Information (Data Types & Missing Values) ---")
print(df.info())

print("\n--- Statistical Summary ---")
print(df.describe())

# -------------------------------------------------------------
# 3. Visualizations (Exploratory Data Analysis)
# -------------------------------------------------------------

# Plot 1: Distribution of Packages (Histogram)
plt.figure(figsize=(8, 5))
sns.histplot(df['Package'], kde=True, color='blue', bins=30)
plt.title('Distribution of Placement Packages (LPA)')
plt.xlabel('Package (LPA)')
plt.ylabel('Count of Students')
plt.savefig('../assets/package_distribution.png') # Saves to your assets folder
plt.show()

# Plot 2: Feature Correlation Matrix (Heatmap)
plt.figure(figsize=(10, 8))
# Compute correlation matrix
correlation_matrix = df.corr()
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt=".2f", linewidths=0.5)
plt.title('Feature Correlation Matrix')
plt.tight_layout()
plt.savefig('../assets/correlation_matrix.png')
plt.show()

# Plot 3: Specific Scatter Analysis (e.g., CGPA vs Package)
plt.figure(figsize=(8, 5))
sns.scatterplot(x='CGPA', y='Package', data=df, alpha=0.6, color='darkgreen')
plt.title('Impact of CGPA on Placement Package')
plt.xlabel('CGPA')
plt.ylabel('Package (LPA)')
plt.savefig('../assets/cgpa_vs_package.png')
plt.show()
```
